# Student Information Dictionary

This project uses a Python dictionary to store student information
and perform search, add, and update operations.

The project demonstrates structured data representation using
unique student IDs as dictionary keys.

```python
# Student Information Dictionary

students = {
    "S001": {
        "name": "Aditi",
        "age": 20,
        "class": "B.Tech",
        "grade": "A"
    },
    "S002": {
        "name": "Rohit",
        "age": 21,
        "class": "B.Sc",
        "grade": "B"
    },
    "S003": {
        "name": "Neha",
        "age": 19,
        "class": "B.Tech",
        "grade": "A+"
    },
    "S004": {
        "name": "Karan",
        "age": 22,
        "class": "BCA",
        "grade": "B"
    },
    "S005": {
        "name": "Priya",
        "age": 20,
        "class": "B.Sc",
        "grade": "A"
    }
}

print("Initial Student Records:")
print(students)
```

```python
def display_students():
    print("===== All Student Records =====")
    
    for student_id, details in students.items():
        print(f"\nStudent ID: {student_id}")
        print(f"Name: {details['name']}")
        print(f"Age: {details['age']}")
        print(f"Class: {details['class']}")
        print(f"Grade: {details['grade']}")


display_students()
```

```python
def search_student(student_id):
    if student_id in students:
        print(f"Student Found: {student_id}")
        print(students[student_id])
    else:
        print("Student not found.")
```

```python
search_student("S003")
```

```python
search_student("S0078")
```

```python
def add_student(student_id, name, age, student_class, grade):
    if student_id in students:
        print("Student ID already exists.")
    else:
        students[student_id] = {
            "name": name,
            "age": age,
            "class": student_class,
            "grade": grade
        }
        print(f"Student {student_id} added successfully.")
```

```python
add_student("S006", "Vikram", 21, "BBA", "B+")
```

```python
search_student("S006")
```

```python
def update_student(student_id, field, new_value):
    if student_id in students:
        if field in students[student_id]:
            students[student_id][field] = new_value
            print(f"{field} updated successfully for {student_id}.")
        else:
            print("Invalid field.")
    else:
        print("Student not found.")
```

```python
update_student("S006", "age", 22)
```

```python
update_student("S002", "grade", "A")
```

```python
search_student("S002")
```

```python
display_students()
```
